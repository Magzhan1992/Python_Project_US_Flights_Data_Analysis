```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
days = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']
temp = [15, 17, 16, 18, 20, 19, 21]

plt.plot(days, temp, marker='o')
plt.xlabel('Дни недели')
plt.ylabel('Температура (°C)')
plt.title('Температура за неделю')
plt.grid()
plt.show()
```

```python
hours = [1, 2, 3, 4, 5, 6, 7]
grades = [2, 3, 3, 4, 4, 5, 5]

plt.scatter(hours, grades)
plt.xlabel('Часы учебы')
plt.ylabel('Оценки')
plt.title('Связь учебы и оценок')
plt.show()
```

```python
sports = ['Футбол', 'Баскетбол', 'Теннис', 'Плавание']
participants = [20, 15, 10, 12]

plt.bar(sports, participants)
plt.xlabel('Виды спорта')
plt.ylabel('Количество участников')
plt.title('Участники по видам спорта')
plt.show()
```

```python
grades = np.random.randint(2, 6, 50)

plt.hist(grades, bins=4)
plt.xlabel('Оценки')
plt.ylabel('Частота')
plt.title('Распределение оценок')
plt.show()
```

```python
coffee = ['Эспрессо', 'Латте', 'Капучино']
coffee_sales = [30, 45, 40]

snacks = ['Круассан', 'Багет', 'Печенье', 'Маффин', 'Сэндвич']
snack_sales = [20, 15, 25, 18, 22]

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.bar(coffee, coffee_sales)
plt.title('Продажи кофе')

plt.subplot(1, 2, 2)
plt.bar(snacks, snack_sales)
plt.title('Продажи закусок')

plt.tight_layout()
plt.show()
```
