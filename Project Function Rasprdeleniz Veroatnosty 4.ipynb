#1. Вероятность того, что сегодня будет дождь составляет 30%, а вероятность того, что ваш заказ на зонт прибудет вовремя, составляет 40%. Вам не терпится прогуляться сегодня под дождем, и вы не можете без него!

Какова вероятность того, что пойдет дождь и ваш зонт прибудет вовремя?

```python
P_rain = 0.30
P_umbrella = 0.40

P_both = P_rain * P_umbrella

print(f"Вероятность дождя и своевременной доставки зонта: {P_both:.0%}")
```

#2. У вас есть закрытый ящик с 10 красными, 15 зелеными и 6 черными шарами. Если мы достанем 3 шара из коробки вслепую, какая вероятность, что первый будет красным, второй зеленым, третий черным?

```python
P_red = 10 / 31
P_green = 15 / 30
P_black = 6 / 29

P_sequence = P_red * P_green * P_black

print(f"Вероятность: {P_sequence:.4f}")
print(f"В процентах: {P_sequence:.2%}")
```

```python
#1.Красный из 31 шара	10/31
2	Красный убран, осталось 30 шаров, зелёных всё ещё 15	15/30
3	Осталось 29 шаров, чёрных всё ещё 6	6/29
Расчёт вручную

P = (10 · 15 · 6) / (31 · 30 · 29) = 900 / 26 970 = 30 / 899 ≈ 0,0334
```

#3. Выполните эксперимент используя binom.rvs со 100 честными монетками, и 1.000 повторениями (size). Сохраните результат в results, и заметьте что если вы не указываете random_state то вы получите разные наборы данных.

```python
from scipy.stats import binom

results = binom.rvs(
    n=100,
    p=0.5,
    size=1000,
    random_state=42
)

print("Первые 10 результатов:", results[:10])
print("Количество экспериментов:", len(results))
print("Среднее количество орлов:", results.mean())
print("Стандартное отклонение:", results.std())
```

#4. Покажите results на гистограмме

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hist(results, bins=15, edgecolor="black")

plt.title("Распределение количества орлов при 100 подбрасываниях")
plt.xlabel("Количество орлов")
plt.ylabel("Частота")
plt.grid(axis="y", alpha=0.3)

plt.show()
```
